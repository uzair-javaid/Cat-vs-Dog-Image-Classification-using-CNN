```python


from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
import matplotlib.pyplot as plt


# Load the Trained Model

model = load_model('cat_dog_classifier.h5')
print("✅ Model loaded successfully!")


# Define Prediction Function

def predict_image(img_path):
    img = load_img(img_path, target_size=(150,150))
    img_array = img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    prediction = model.predict(img_array)[0][0]

    # Display image
    plt.imshow(load_img(img_path))
    plt.axis('off')

    if prediction > 0.5:
        plt.title(f" Dog (Confidence: {prediction:.2f})", color='green')
        print(f" Prediction: It's a Dog! (Confidence: {prediction:.4f})")
    else:
        plt.title(f" Cat (Confidence: {1 - prediction:.2f})", color='blue')
        print(f" Prediction: It's a Cat! (Confidence: {1 - prediction:.4f})")

    plt.show()


# Test on Any Image

predict_image('dataset/test/cats/cat.4035.jpg')

```
